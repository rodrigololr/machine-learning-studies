# Fase 2: Preprocessamento

## Objetivo
Normalizar/escalar os dados biometricos usando:
- StandardScaler (z-score)
- MinMaxScaler (0-1)

Preparar dados para algoritmos de clustering.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import pickle

OUTPUT_DIR = '/home/antonio/Documentos/Machine Learning/atividade_3'
numeric_cols = ['AL', 'ACD', 'WTW', 'K1', 'K2']

# Carregar dados
df = pd.read_pickle(f'{OUTPUT_DIR}/df_clean.pkl')
X = df[numeric_cols].values
print(f'Dados: {X.shape}')
```

## StandardScaler (Z-Score)

```python
# StandardScaler
scaler_std = StandardScaler()
X_std = scaler_std.fit_transform(X)

print('StandardScaler (Z-Score):')
print(f'  Media: {X_std.mean(axis=0).round(6)}')
print(f'  Std: {X_std.std(axis=0).round(6)}')

# Salvar scaler e dados
with open(f'{OUTPUT_DIR}/scaler_standard.pkl', 'wb') as f:
    pickle.dump(scaler_std, f)

df_std = pd.DataFrame(X_std, columns=numeric_cols)
df_std.to_pickle(f'{OUTPUT_DIR}/X_standard.pkl')
print('\nSalvo: scaler_standard.pkl, X_standard.pkl')
```

## MinMaxScaler (0-1)

```python
# MinMaxScaler
scaler_minmax = MinMaxScaler()
X_minmax = scaler_minmax.fit_transform(X)

print('MinMaxScaler (0-1):')
print(f'  Min: {X_minmax.min(axis=0).round(6)}')
print(f'  Max: {X_minmax.max(axis=0).round(6)}')

# Salvar scaler e dados
with open(f'{OUTPUT_DIR}/scaler_minmax.pkl', 'wb') as f:
    pickle.dump(scaler_minmax, f)

df_minmax = pd.DataFrame(X_minmax, columns=numeric_cols)
df_minmax.to_pickle(f'{OUTPUT_DIR}/X_minmax.pkl')
print('\nSalvo: scaler_minmax.pkl, X_minmax.pkl')
```

## Comparacao dos Metodos

```python
import matplotlib.pyplot as plt

# Comparar distribuicoes antes e depois
fig, axes = plt.subplots(2, 5, figsize=(18, 8))

for idx, col in enumerate(numeric_cols):
    # Original
    axes[0, idx].hist(X[:, idx], bins=30, alpha=0.7, color='steelblue')
    axes[0, idx].set_title(f'{col} - Original')
    
    # Standard
    axes[1, idx].hist(X_std[:, idx], bins=30, alpha=0.7, color='green')
    axes[1, idx].set_title(f'{col} - Standard')
```

```python
plt.suptitle('Comparacao: Original vs StandardScaler', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/imgs/preprocessing_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Salvo: {OUTPUT_DIR}/imgs/preprocessing_comparison.png')
```

```python
# Resumo
print('\n=== Resumo Preprocessamento ===')
print(f'Dados originais: shape={X.shape}')
print(f'X_standard: shape={X_std.shape}, mean~0, std~1')
print(f'X_minmax: shape={X_minmax.shape}, range=[0,1]')
print('\nArquivos salvos:')
print('  - scaler_standard.pkl')
print('  - X_standard.pkl')
print('  - scaler_minmax.pkl')
print('  - X_minmax.pkl')
```
