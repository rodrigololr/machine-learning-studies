# Fase 6: Hierarchical Clustering

## Objetivo
Aplicar clustering hierarquico com Ward linkage e visualizar com dendrograma.

```python
import pandas as pd
import numpy as np
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

OUTPUT_DIR = '/home/antonio/Documentos/Machine Learning/atividade_3'
IMG_DIR = f'{OUTPUT_DIR}/imgs'

# Carregar dados
X_std = pd.read_pickle(f'{OUTPUT_DIR}/X_standard.pkl').values
print(f'Dados: {X_std.shape}')
```

```python
# Hierarchical clustering com Ward linkage
linkage_matrix = linkage(X_std, method='ward')
print('Linkage matrix computed')
```

```python
# Dendrograma
fig, ax = plt.subplots(figsize=(16, 8))
dendrogram(linkage_matrix, ax=ax, truncate_mode='lastp', p=50, 
           leaf_rotation=90, leaf_font_size=10, show_contracted=True)
ax.set_title('Dendrograma - Ward Linkage', fontsize=14, fontweight='bold')
ax.set_xlabel('Amostra ou Cluster ID', fontsize=12)
ax.set_ylabel('Distancia (Ward)', fontsize=12)
ax.axhline(y=50, color='r', linestyle='--', label='Corte para 2 clusters')
ax.legend()
plt.tight_layout()
plt.savefig(f'{IMG_DIR}/dendrogram.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Aplicar hierarchical com 2 clusters
hc = AgglomerativeClustering(n_clusters=2, linkage='ward')
hc_labels = hc.fit_predict(X_std)
print(f'\nHierarchical (2 clusters):')
print(f'  Cluster sizes: {np.bincount(hc_labels)}')
```

```python
# Salvar labels
np.save(f'{OUTPUT_DIR}/hierarchical_labels.npy', hc_labels)
print(f'Labels salvos')
```
