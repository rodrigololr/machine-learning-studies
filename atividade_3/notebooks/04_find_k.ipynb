# Fase 3: Encontrar K Otimo

## Metodos
1. **Elbow Method**: Plot inertia (SSE) vs K. O 'cotovelo' indica o K otimo.
2. **Silhouette Analysis**: Mede quao bem pontos pertencem ao cluster. Maior = melhor.

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

OUTPUT_DIR = '/home/antonio/Documentos/Machine Learning/atividade_3'
IMG_DIR = f'{OUTPUT_DIR}/imgs'

# Carregar dados padronizados
X_std = pd.read_pickle(f'{OUTPUT_DIR}/X_standard.pkl').values
print(f'Dados: {X_std.shape}')
```

```python
# Testar K de 2 a 15
k_range = range(2, 16)
inertias = []
silhouettes = []

print('Calculando K from 2 to 15...')
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_std)
    inertias.append(kmeans.inertia_)
    sil = silhouette_score(X_std, kmeans.labels_)
    silhouettes.append(sil)
    print(f'  K={k}: inertia={kmeans.inertia_:.2f}, silhouette={sil:.4f}')
```

```python
# Elbow plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(k_range, inertias, 'bo-', linewidth=2, markersize=8)
ax.set_xlabel('K (numero de clusters)', fontsize=12)
ax.set_ylabel('Inertia (Soma das distancias quadraticas)', fontsize=12)
ax.set_title('Metodo do Cotovelo (Elbow Method)', fontsize=14, fontweight='bold')
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f'{IMG_DIR}/elbow_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Silhouette plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(k_range, silhouettes, 'go-', linewidth=2, markersize=8)
ax.set_xlabel('K (numero de clusters)', fontsize=12)
ax.set_ylabel('Silhouette Score', fontsize=12)
ax.set_title('Analise de Silhueta', fontsize=14, fontweight='bold')
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3)
best_k = list(k_range)[np.argmax(silhouettes)]
ax.axvline(x=best_k, color='red', linestyle='--', label=f'Melhor K={best_k}')
ax.legend()
plt.tight_layout()
plt.savefig(f'{IMG_DIR}/silhouette_plot.png', dpi=150, bbox_inches='tight')
plt.show()

print(f'\nMelhor K: {best_k} com silhouette={max(silhouettes):.4f}')
```
