# Fase 4: K-Means Clustering

## Objetivo
Aplicar K-Means com K=2 (encontrado na fase anterior) e visualizar resultados.

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

OUTPUT_DIR = '/home/antonio/Documentos/Machine Learning/atividade_3'
IMG_DIR = f'{OUTPUT_DIR}/imgs'

# Carregar dados
X_std = pd.read_pickle(f'{OUTPUT_DIR}/X_standard.pkl').values
print(f'Dados: {X_std.shape}')
```

```python
# K-Means com K=2
optimal_k = 2
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_std)
print(f'K-Means com K={optimal_k}')
print(f'Cluster sizes: {np.bincount(labels)}')
```

```python
# PCA visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_std)
print(f'PCA explained variance: {pca.explained_variance_ratio_.sum():.3f}')
```

```python
# t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_std)
print('t-SNE done')
```

```python
# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

scatter = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', alpha=0.6, s=30)
axes[0].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
                 c='red', marker='X', s=200, edgecolors='black', linewidths=2, label='Centroides')
axes[0].set_xlabel('PC1')
axes[0].set_ylabel('PC2')
axes[0].set_title(f'K-Means (K={optimal_k}) - PCA', fontweight='bold')
axes[0].legend()
plt.colorbar(scatter, ax=axes[0], label='Cluster')

scatter = axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='viridis', alpha=0.6, s=30)
axes[1].set_xlabel('t-SNE 1')
axes[1].set_ylabel('t-SNE 2')
axes[1].set_title(f'K-Means (K={optimal_k}) - t-SNE', fontweight='bold')
plt.colorbar(scatter, ax=axes[1], label='Cluster')

plt.tight_layout()
plt.savefig(f'{IMG_DIR}/kmeans_pca_tsne.png', dpi=150, bbox_inches='tight')
plt.show()
```
